# 02 — Reproduce external validation

The five trained fold checkpoints and external ESM-2 features are distributed with the repository. This notebook therefore runs directly after the TensorFlow environment is installed.

```python
from pathlib import Path
import subprocess, sys, json
import pandas as pd
ROOT = Path.cwd()
if not (ROOT / "src").exists(): ROOT = ROOT.parent
print("Repository root:", ROOT)
```

```python
cmd = [sys.executable, str(ROOT / "external_validation/run_external_validation.py")]
print(" ".join(cmd))
subprocess.run(cmd, cwd=ROOT, check=True)
```
